```python
# Generates statistical summaries and visualizations for the cleaned
# heart disease dataset, including distributions, outlier detection,
# feature importance, and a correlation heatmap.
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# Load Cleaned Dataset
df=pd.read_csv('C:/Users/Win 11/Desktop/python/IpProject2/data/cleaned_data.csv')
# Display descriptive statistics for all numerical columns
print(df.describe())
```

```python
# Age Distribution
# Histogram showing how patient ages are distributed in the dataset
plt.hist(df["age"],bins=8,color='blue',alpha=0.5)
plt.title("Age Distribution ")
plt.xlabel("age")
plt.ylabel("count")
plt.show()
```

```python
# Cholesterol Distribution
# Histogram showing the spread of cholesterol levels across patients
plt.hist(df["chol"],bins=8,color='blue',alpha=0.5)
plt.title("chol Distribution ")
plt.xlabel("chol")
plt.ylabel("count")
plt.show()
```

```python
# Blood Pressure Outlier Detection
# Boxplot to identify outliers in resting blood pressure values
plt.boxplot(df["trestbps"])
plt.title("Resting Blood Pressure Distribution by Target Class")
plt.ylabel("trestbps")
plt.show()
```

```python
# Max Heart Rate Outlier Detection
# Boxplot to detect anomalies in maximum heart rate achieved
plt.boxplot(df["thalach"])
plt.title("Detection of Outliers in Maximum Heart Rate (thalach)")
plt.ylabel("thalach")
plt.show()
```

```python
# Feature Importance Ranking
# Bar chart displaying manually assigned importance scores for key features
feature=['age','trestbps','chol','thalach','oldpeak']
plt.title("Feature Importance Ranking for Heart Disease Prediction")
importannce_scores=[ 0.15,0.30,0.45,0.60,0.85]
bars=plt.bar(feature,importannce_scores,color='purple',alpha=0.5)
```

```python
# Correlation Heatmap
# Heatmap showing pairwise correlations between key features and the target
coulms= df[['age', 'trestbps', 'thalach', 'oldpeak', 'target']]
corr_matrix = coulms.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")

plt.title("Feature Correlation Heatmap")
plt.show()
```
